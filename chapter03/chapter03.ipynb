```python
from openai import OpenAI

client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

# To see how a specific model formats things, 
# we use the /completions endpoint (Legacy) but point it at a Chat model.
response = client.completions.create(
    model="gemma4:e4b",
    prompt="This is a test", 
    echo=True, # This tells the server to return the prompt + the result
    max_tokens=0, # We don't want a response, just the prompt
)

print(f"Native Format:\n{response.choices[0].text}")
```

```python
from openai import OpenAI

client = OpenAI(base_url="http://localhost:8080/v1", api_key="sk-local")

response = client.chat.completions.create(
    model="local-model",
    messages=[
        {"role": "system", "content": "You are a teacher."},
        {"role": "user", "content": "Explain math."}
    ],
    # Non-standard: Tells llama-server to include the prompt string in the JSON response
    extra_body={
        "image_data": [], 
        "samplers": []
    }
)

# Note: In llama-server, the raw prompt is often printed to the TERMINAL 
# where the server is running, but you can also find it in the 'usage' or 
# hidden fields of the response object depending on the version.
```
